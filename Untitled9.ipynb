STEP 1

We first install PySpark and Delta Spark and create a SparkSession configured with Delta Lake support

```python
!pip install -q pyspark delta-spark

import pyspark
from delta import *
from pyspark.sql import SparkSession
from pyspark.sql.functions import col, when, to_date, count, sum

builder = (
    SparkSession.builder.appName("FraudDetectionMedallion")
    .master("local[*]")
    .config("spark.sql.extensions", "io.delta.sql.DeltaSparkSessionExtension")
    .config(
        "spark.sql.catalog.spark_catalog",
        "org.apache.spark.sql.delta.catalog.DeltaCatalog",
    )
)

spark = configure_spark_with_delta_pip(builder).getOrCreate()

print(" Spark Session successfully initialized!")
```

STEP 2

Now we read the raw CSV files into Spark DataFrames and save them as raw Delta Tables WHICH IS BASICALLY the Bronze layer
We write the raw data into the Bronze delta tables
and preview it

equals signs used for visual decoration and formatting.

```python
bronze_acc_df = spark.read.csv("accounts.csv", header=True, inferSchema=True)
bronze_txns_df = spark.read.csv(
    "transactions.csv", header=True, inferSchema=True
)
bronze_fraud_df = spark.read.csv(
    "known_fraud_accounts.csv", header=True, inferSchema=True
)

bronze_acc_df.write.format("delta").mode("overwrite").save(
    "/content/delta/bronze/accounts"
)
bronze_txns_df.write.format("delta").mode("overwrite").save(
    "/content/delta/bronze/transactions"
)
bronze_fraud_df.write.format("delta").mode("overwrite").save(
    "/content/delta/bronze/fraud_watchlist"
)

print("--- Bronze Accounts Preview ---")
bronze_acc_df.show(5)
```

STEP 3

Now we read from the Bronze Delta tables (not the raw CSVs) and clean and organise it to handle the NULL valus format dates etc.

```python
bronze_acc = spark.read.format("delta").load("/content/delta/bronze/accounts")
bronze_txns = spark.read.format("delta").load(
    "/content/delta/bronze/transactions"
)

clean_acc = (
    bronze_acc.filter(col("account_id").isNotNull())
    .withColumn("credit_limit", col("credit_limit").cast("double"))
    .fillna({"credit_limit": 0.0})
    .withColumn("opening_date", to_date(col("opening_date")))
)

clean_txns = (
    bronze_txns.filter(col("account_id").isNotNull())
    .withColumn("amount", col("amount").cast("double"))
    .withColumn("txn_date", to_date(col("txn_date")))
)

silver_enriched_df = clean_txns.join(
    clean_acc, on="account_id", how="inner"
).select(
    clean_txns["txn_id"],
    clean_txns["account_id"],
    clean_acc["customer_name"],
    clean_acc["account_type"],
    clean_acc["branch"],
    clean_acc["kyc_status"],
    clean_acc["credit_limit"],
    clean_txns["txn_date"],
    clean_txns["amount"],
    clean_txns["merchant"],
)

silver_enriched_df.write.format("delta").mode("overwrite").save(
    "/content/delta/silver/enriched_transactions"
)

print("--- Silver Enriched Transactions Preview ---")
silver_enriched_df.show(5)
print(f"Total Enriched Rows: {silver_enriched_df.count()}")
```

STEP 4

HERE we load the silver tranasctions and bronze watchlist.
then perfrom a left join on accounts id
to identify the fraud and non fraud ones.


```python
silver_txns = spark.read.format("delta").load(
    "/content/delta/silver/enriched_transactions"
)
bronze_fraud = spark.read.format("delta").load(
    "/content/delta/bronze/fraud_watchlist"
)

gold_classified_df = silver_txns.join(
    bronze_fraud.select("account_id", "fraud_type", "flagged_date"),
    on="account_id",
    how="left",
).withColumn(
    "status",
    when(col("fraud_type").isNotNull(), "fraud").otherwise("normal"),
)

gold_classified_df.write.format("delta").mode("overwrite").save(
    "/content/delta/gold/classified_transactions"
)

print("--- Gold Layer Classified Transactions Preview ---")
gold_classified_df.select(
    "txn_id",
    "account_id",
    "customer_name",
    "amount",
    "fraud_type",
    "status",
).show(10)
```

STEP 5

To confirm thAt the  fraud logic matched correctly, we add a new cell to filter and view the detected fraud transactions:

```python
print("--- Flagged Fraud Transactions ---")
gold_classified_df.filter(col("status") == "fraud").select(
    "txn_id",
    "account_id",
    "customer_name",
    "amount",
    "fraud_type",
    "status",
).show(10)
```

STEP 6

Spark SQL Aggregations & Metrics

```python
gold_classified_df.createOrReplaceTempView("gold_transactions")

print("================== 1. OVERALL SUMMARY ==================")
spark.sql(
    """
    SELECT
        COUNT(txn_id) AS total_transactions,
        SUM(CASE WHEN status = 'fraud' THEN 1 ELSE 0 END) AS total_fraud_transactions,
        SUM(CASE WHEN status = 'normal' THEN 1 ELSE 0 END) AS total_normal_transactions,
        ROUND(SUM(CASE WHEN status = 'fraud' THEN amount ELSE 0 END), 2) AS total_fraud_amount,
        ROUND(SUM(amount), 2) AS total_volume_amount
    FROM gold_transactions
"""
).show()


print("================ 2. FRAUD TYPE BREAKDOWN ================")
spark.sql(
    """
    SELECT
        fraud_type,
        COUNT(txn_id) AS fraud_count,
        ROUND(SUM(amount), 2) AS total_amount,
        ROUND(AVG(amount), 2) AS avg_transaction_amount
    FROM gold_transactions
    WHERE status = 'fraud'
    GROUP BY fraud_type
    ORDER BY fraud_count DESC
"""
).show()


print("============ 3. ACCOUNT-LEVEL FRAUD ANALYSIS ============")
spark.sql(
    """
    SELECT
        account_id,
        customer_name,
        branch,
        fraud_type,
        COUNT(txn_id) AS fraud_txns_count,
        ROUND(SUM(amount), 2) AS total_fraud_amount
    FROM gold_transactions
    WHERE status = 'fraud'
    GROUP BY account_id, customer_name, branch, fraud_type
    ORDER BY total_fraud_amount DESC
"""
).show()
```

EXTRA

Multi-Factor Dynamic Fraud Risk Scoring Engine (0–100 Score)

I ma adding this to provide an extra touch to my project.

While a static watchlist will only  detect accounts that have already been caught, this additonal engine which i added evaluates multi-dimensional behavioral anomalies (credit limit breaches, unverified KYC exposure, and high-value activity) on a continuous 0–100 risk scale.

This transforms the pipeline from a passive historical reporting tool into an active fraud prevention engine, automatically prioritizing alerts into actionable tiers (LOW, MEDIUM, HIGH, and CRITICAL).



```python
from pyspark.sql.functions import col, lit, when

gold_scored_df = (
    gold_classified_df.withColumn(
        "score_watchlist",
        when(col("fraud_type").isNotNull(), 40).otherwise(0),
    )
    .withColumn(
        "score_limit_breach",
        when(
            (col("credit_limit") > 0) & (col("amount") > col("credit_limit")),
            30,
        ).otherwise(0),
    )
    .withColumn(
        "score_kyc_risk",
        when(
            col("kyc_status").isin(["pending", "expired"])
            & (col("amount") > 50000),
            20,
        ).otherwise(0),
    )
    .withColumn(
        "score_high_value", when(col("amount") > 100000, 10).otherwise(0)
    )
)

gold_scored_df = (
    gold_scored_df.withColumn(
        "total_risk_score",
        col("score_watchlist")
        + col("score_limit_breach")
        + col("score_kyc_risk")
        + col("score_high_value"),
    )
    .withColumn(
        "total_risk_score",
        when(col("total_risk_score") > 100, 100).otherwise(
            col("total_risk_score")
        ),
    )
    .withColumn(
        "risk_tier",
        when(col("total_risk_score") >= 70, "CRITICAL")
        .when(col("total_risk_score") >= 40, "HIGH")
        .when(col("total_risk_score") >= 20, "MEDIUM")
        .otherwise("LOW"),
    )
)

gold_scored_df.write.format("delta").mode("overwrite").save(
    "/content/delta/gold/scored_transactions"
)

print("================== RISK TIER SUMMARY ==================")
gold_scored_df.groupBy("risk_tier").count().orderBy(
    col("count").desc()
).show()

print("============= CRITICAL RISK TRANSACTIONS =============")
gold_scored_df.filter(col("risk_tier") == "CRITICAL").select(
    "txn_id",
    "account_id",
    "customer_name",
    "amount",
    "credit_limit",
    "fraud_type",
    "total_risk_score",
    "risk_tier",
).show(truncate=False)
```
